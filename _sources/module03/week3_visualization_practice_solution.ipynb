## 1. Set up and inspect the data

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

LOCAL_DATA_PATH = Path("../data/penguins.csv")
DATA_URL = (
    "https://raw.githubusercontent.com/"
    "obscrivn/DataScience-book/main/data/penguins.csv"
)

data_source = LOCAL_DATA_PATH if LOCAL_DATA_PATH.exists() else DATA_URL
penguins = pd.read_csv(data_source, na_values=["NA"])

sns.set_theme(style="whitegrid", context="notebook")

print(f"Loaded {penguins.shape[0]} rows and {penguins.shape[1]} columns.")
print(f"Data source: {data_source}")
penguins.head()
```

## 2. One variable, two useful views

A histogram and a boxplot can examine the same numeric variable, but they summarize it differently.

**Predict:** Before running the next cell, where do you expect most body-mass observations to fall? Do you expect one smooth peak or more than one concentration?

```python
mass_data = penguins.dropna(subset=["body_mass_g"]).copy()

print(
    f"Body-mass plot uses {len(mass_data)} of {len(penguins)} rows "
    "(rows missing body_mass_g are excluded only from these plots)."
)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.histplot(
    data=mass_data,
    x="body_mass_g",
    bins=18,
    color="#0072B2",
    edgecolor="white",
    ax=axes[0],
)
axes[0].set(
    title="Distribution of penguin body mass",
    xlabel="Body mass (g)",
    ylabel="Number of penguins",
)

sns.boxplot(
    data=mass_data,
    x="body_mass_g",
    color="#56B4E9",
    ax=axes[1],
)
axes[1].set(
    title="Compact summary of body mass",
    xlabel="Body mass (g)",
    ylabel="",
)

fig.tight_layout()
plt.show()
```

### Worked interpretation

The histogram reveals multiple concentrations in the combined body-mass distribution. The boxplot more compactly shows the median, middle half, range, and potential extreme values, but it hides the detailed shape.

Neither plot explains *why* the distribution has this structure. A useful next step is to examine a relevant grouping variable. Also remember the Week 02 principle: an unusual point is a reason to investigate, not an automatic deletion.

**Possible response:** I expected one broad concentration, but the histogram shows multiple concentrations. The compact boxplot does not make that structure nearly as visible, so grouping by species is a useful next step.

## 3. Grouped comparison: distribution and aggregation

The question now changes from “What is the overall distribution?” to “How does body mass vary across species?”

The two charts below answer related but different questions:

- A **distribution view** shows within-species spread, overlap, median, and possible unusual values.
- An **aggregated view** makes the mean easy to compare, but compresses each species to one number.

Neither view is universally superior. Choose based on the analytical question, and explain what the chosen summary leaves out.

```python
SPECIES_ORDER = ["Adelie", "Chinstrap", "Gentoo"]
SPECIES_COLORS = ["#0072B2", "#D55E00", "#009E73"]

grouped_data = penguins.dropna(subset=["species", "body_mass_g"]).copy()
group_summary = (
    grouped_data.groupby("species", observed=True)["body_mass_g"]
    .agg(observations="size", mean_g="mean", median_g="median", std_g="std")
    .reindex(SPECIES_ORDER)
    .round(1)
)
display(group_summary)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

sns.boxplot(
    data=grouped_data,
    x="species",
    y="body_mass_g",
    order=SPECIES_ORDER,
    color="#56B4E9",
    ax=axes[0],
)
axes[0].set(
    title="Body-mass distributions by species",
    xlabel="Species",
    ylabel="Body mass (g)",
)

axes[1].bar(
    group_summary.index,
    group_summary["mean_g"],
    color=SPECIES_COLORS,
)
axes[1].set(
    title="Mean body mass by species",
    xlabel="Species",
    ylabel="Mean body mass (g)",
)
axes[1].set_ylim(0, group_summary["mean_g"].max() * 1.15)

for index, value in enumerate(group_summary["mean_g"]):
    axes[1].text(index, value + 75, f"{value:,.0f}", ha="center")

fig.tight_layout()
plt.show()
```

### Compare, then modify

1. Which chart better supports a question about individual variation and overlap?
2. Which chart makes the group means easiest to compare?
3. What information is hidden by each chart?
4. Do the unequal species counts affect how confidently you read the comparison?

**Possible response:**

1. The boxplots better support questions about within-species variation, medians, and overlap.
2. The bar chart makes the three group means easiest to compare directly.
3. The boxplots do not emphasize the exact means, while the mean bars hide distribution shape, overlap, and unusual observations.
4. Yes. The group summary shows 151 Adelie, 68 Chinstrap, and 123 Gentoo observations with body mass. The smaller Chinstrap group provides less information about its population distribution.

The solution below demonstrates the guided modification with `bill_depth_mm`.

```python
guided_measure = "bill_depth_mm"
guided_label = "Bill depth (mm)"

guided_data = penguins.dropna(subset=["species", guided_measure]).copy()

fig, ax = plt.subplots(figsize=(7, 4))
sns.boxplot(
    data=guided_data,
    x="species",
    y=guided_measure,
    order=SPECIES_ORDER,
    color="#E69F00",
    ax=ax,
)
ax.set(
    title=f"{guided_label} by species",
    xlabel="Species",
    ylabel=guided_label,
)
fig.tight_layout()
plt.show()

print(f"This comparison uses {len(guided_data)} of {len(penguins)} rows.")
```

### Guided interpretation

**Possible response:** Adelie and Chinstrap bill-depth distributions overlap substantially and have similar medians, while Gentoo measurements are generally lower and overlap less with the other two groups. The boxplots reveal within-species spread, overlap, medians, and possible unusual observations that a table of means alone would compress.

## 4. Two numeric variables with a grouping variable

A scatterplot fits a question about association between two numeric variables. Species adds a third variable.

This plot uses both color and marker shape for species. That redundant encoding is a lightweight accessibility improvement: readers do not have to rely on color alone. Smaller, partly transparent markers also reduce overplotting.

```python
species_palette = {
    "Adelie": "#0072B2",
    "Chinstrap": "#D55E00",
    "Gentoo": "#009E73",
}
species_markers = {
    "Adelie": "o",
    "Chinstrap": "s",
    "Gentoo": "^",
}

relationship_data = penguins.dropna(
    subset=["flipper_length_mm", "body_mass_g", "species"]
).copy()

fig, ax = plt.subplots(figsize=(8, 5.5))
sns.scatterplot(
    data=relationship_data,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    style="species",
    hue_order=SPECIES_ORDER,
    style_order=SPECIES_ORDER,
    palette=species_palette,
    markers=species_markers,
    s=60,
    alpha=0.72,
    ax=ax,
)
ax.set(
    title="Flipper length and body mass by species",
    xlabel="Flipper length (mm)",
    ylabel="Body mass (g)",
)
ax.legend(title="Species", frameon=True)
fig.tight_layout()
plt.show()

overall_correlation = relationship_data[
    ["flipper_length_mm", "body_mass_g"]
].corr().iloc[0, 1]
print(f"Overall Pearson correlation: {overall_correlation:.3f}")
print(f"Plot uses {len(relationship_data)} of {len(penguins)} rows.")
```

### Interpret without overclaiming

**Possible response:** Flipper length and body mass have a strong positive overall association in the complete observations (Pearson correlation about 0.87). Species helps explain the visible clusters, especially the longer-flippered, heavier Gentoo observations. The chart does not show that increasing a penguin's flipper length would cause it to gain mass; a useful next step would compare the relationship within each species and investigate other variables such as sex or age.

## 5. Critique a simulated AI-generated draft

Imagine an AI coding assistant returned the chart below with this claim:

> “Species determines body mass, and every Gentoo penguin is heavier than every Adelie or Chinstrap penguin.”

This is a **simulated draft**. No live AI connection is used. Run the code, then audit the data, code, chart, and claim before trusting it.

```python
ai_summary = (
    grouped_data.groupby("species", observed=True)["body_mass_g"]
    .mean()
    .reindex(SPECIES_ORDER)
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(ai_summary.index, ai_summary.values, color=SPECIES_COLORS)
ax.set_ylim(ai_summary.min() - 200, ai_summary.max() + 200)
ax.set(
    title="Species determines penguin body mass",
    xlabel="Species",
    ylabel="Mean body mass (g)",
)
fig.tight_layout()
plt.show()
```

### Audit before repair

**Possible response:**

1. **Data:** No. A mean summarizes a group and does not describe every individual observation.
2. **Code:** The code groups rows by species and replaces each group's body-mass distribution with one mean.
3. **Chart:** Starting the y-axis near the smallest mean exaggerates the apparent separation between bars.
4. **Claim:** “Determines” implies causation, while “every” claims certainty that the group ranges contradict.
5. **Accessibility:** The species names on the x-axis preserve the comparison without color, so color is decorative here. A single neutral bar color or direct value labels could make that intent clearer.

```python
claim_check = (
    grouped_data.groupby("species", observed=True)["body_mass_g"]
    .agg(
        observations="size",
        minimum_g="min",
        maximum_g="max",
        mean_g="mean",
        median_g="median",
    )
    .reindex(SPECIES_ORDER)
    .round(1)
)
claim_check
```

### Compare two defensible repairs

The overlap in the minimum and maximum values disproves the word “every.” The next cell shows two possible repairs:

- The boxplot supports a question about distributions and overlap.
- The point-and-error-bar view supports a question about means while retaining a measure of within-group variation.

These are alternatives for different questions—not a ranking in which one chart type always wins.

```python
repair_summary = (
    grouped_data.groupby("species", observed=True)["body_mass_g"]
    .agg(mean_g="mean", std_g="std")
    .reindex(SPECIES_ORDER)
)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

sns.boxplot(
    data=grouped_data,
    x="species",
    y="body_mass_g",
    order=SPECIES_ORDER,
    color="#56B4E9",
    ax=axes[0],
)
axes[0].set(
    title="Body-mass distributions overlap across species",
    xlabel="Species",
    ylabel="Body mass (g)",
)

x_positions = range(len(repair_summary))
axes[1].errorbar(
    x_positions,
    repair_summary["mean_g"],
    yerr=repair_summary["std_g"],
    fmt="o",
    color="#0072B2",
    ecolor="#4D4D4D",
    capsize=5,
    markersize=7,
)
axes[1].set_xticks(list(x_positions), repair_summary.index)
axes[1].set(
    title="Species means with ±1 standard deviation",
    xlabel="Species",
    ylabel="Body mass (g)",
)

fig.tight_layout()
plt.show()
```

### Repair reflection

**Possible response:** The boxplot answers how body-mass distributions differ and overlap across species. It shows that Gentoo observations tend to be heavier, but it still does not establish causation, explain every individual difference, or display the detailed distribution shape as clearly as a jittered point or violin view might.

## 6. Short independent mini-analysis

Spend **no more than 8–10 minutes** on this section. Choose **one** question:

- How does bill depth vary across islands?
- How are bill length and bill depth related, and how does species affect the pattern?

**Solution example:** Use `bill_length_mm` and `bill_depth_mm` as the two numeric variables and `species` as the grouping variable. A scatterplot fits the relationship question and makes species-specific clusters visible.

```python
independent_data = penguins.dropna(
    subset=["bill_length_mm", "bill_depth_mm", "species"]
).copy()

independent_palette = {
    "Adelie": "#0072B2",
    "Chinstrap": "#D55E00",
    "Gentoo": "#009E73",
}
independent_markers = {
    "Adelie": "o",
    "Chinstrap": "s",
    "Gentoo": "^",
}

fig, ax = plt.subplots(figsize=(8, 5.5))
sns.scatterplot(
    data=independent_data,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    style="species",
    hue_order=["Adelie", "Chinstrap", "Gentoo"],
    style_order=["Adelie", "Chinstrap", "Gentoo"],
    palette=independent_palette,
    markers=independent_markers,
    s=60,
    alpha=0.72,
    ax=ax,
)
ax.set(
    title="Bill length and depth form species-specific patterns",
    xlabel="Bill length (mm)",
    ylabel="Bill depth (mm)",
)
ax.legend(title="Species", frameon=True)
fig.tight_layout()
plt.show()

print(f"This analysis uses {len(independent_data)} of {len(penguins)} rows.")
```

### Independent interpretation

**Possible response:**

1. **Evidence:** Species forms distinct clusters: Gentoo penguins tend to have longer, shallower bills, Adelie penguins tend to have shorter, deeper bills, and Chinstrap penguins tend to have longer, deeper bills. Within each species, bill length and depth show a positive association even though the combined association is weakly negative.
2. **Limit:** The plot does not show that bill length causes bill depth, and it does not explain variation from unmeasured characteristics.
3. **Check:** I would calculate the bill-length/bill-depth correlation within each species and compare group counts and numeric summaries before communicating the pattern.

Before sharing any exploratory chart, ask:

- **Question:** Does the chart type match the variables and analytical question?
- **Data:** Are filtering, missing values, grouping, and aggregation visible and justified?
- **Scale:** Are axes and comparisons honest?
- **Access:** Are labels and units readable, and can categories be distinguished without color alone?
- **Evidence:** Does the written claim match what the chart actually shows?
- **Limit:** Have I stated what the visualization cannot establish?
